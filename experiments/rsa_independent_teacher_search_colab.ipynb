```python
import base64,gzip,urllib.request
base='https://raw.githubusercontent.com/hanialshater/LSH_Memory/refs/heads/rsa-v2-colab/experiments/'
blob=''.join(urllib.request.urlopen(base+f'rsa_ind_teacher_small.part{i}').read().decode() for i in range(6))
src=gzip.decompress(base64.b64decode(blob)).decode()
src=src.replace('all_images = df[\"image\"].tolist()', 'all_images = [ds[i][\"image\"] for i in range(len(ds))]')
src=src.replace('        inputs = {k: v.to(device) for k, v in inputs.items()}\n        out = model(**inputs)\n        emb = out.image_embeds\n', '        pixel_values = inputs[\"pixel_values\"].to(device)\n        vision_out = model.vision_model(pixel_values=pixel_values)\n        emb = model.visual_projection(vision_out.pooler_output)\n')
src=src.replace('    inputs = {k: v.to(device) for k, v in inputs.items()}\n    out = model(**inputs)\n    emb = out.text_embeds\n', '    input_ids = inputs[\"input_ids\"].to(device)\n    attention_mask = inputs.get(\"attention_mask\")\n    if attention_mask is not None:\n        attention_mask = attention_mask.to(device)\n    text_out = model.text_model(input_ids=input_ids, attention_mask=attention_mask)\n    emb = model.text_projection(text_out.pooler_output)\n')
exec(src)
```
